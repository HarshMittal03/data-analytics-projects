Task 2.2 — Inspect and Clean the Data

Answer

Data cleaning is an important step in EDA because raw data may contain missing values, duplicate records, incorrect data types, and inconsistent values.

The following cleaning operations are performed:

Check missing values.

Check duplicate records.

Convert date columns into datetime format.

Convert numerical columns into numeric data types.

Fill missing numerical values using the median.

Fill missing text values using "Unknown".

Remove duplicate rows.

Remove duplicate track IDs.

Create useful new columns.

Create a copy of the dataset

```python
df = tracks.copy()

print("Original Shape:", df.shape)
```

Check missing values

```python
missing_values = df.isnull().sum()

display(missing_values)
```

Calculate missing-value percentage

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

display(missing_percentage)
```

Check duplicate rows

```python
duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

Check duplicate track IDs

```python
duplicate_ids = df["id"].duplicated().sum()

print("Number of duplicate track IDs:", duplicate_ids)
```

Convert release date

```python
df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce"
)

print("Release date converted successfully!")
```

Convert numerical columns

```python
numeric_columns = [
    "valence",
    "year",
    "acousticness",
    "danceability",
    "duration_ms",
    "energy",
    "instrumentalness",
    "key",
    "liveness",
    "loudness",
    "mode",
    "popularity",
    "speechiness",
    "tempo"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("Numerical columns converted successfully!")
```

Fill missing numerical values

```python
for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )

print("Missing numerical values handled!")
```

Fill missing text values

```python
df["artists"] = df["artists"].fillna(
    "Unknown Artist"
)

df["name"] = df["name"].fillna(
    "Unknown Track"
)

print("Missing text values handled!")
```

Remove duplicate rows

```python
df = df.drop_duplicates()

print("Duplicate rows removed!")
print("Current Shape:", df.shape)
```

Remove duplicate track IDs

```python
df = df.drop_duplicates(
    subset="id"
)

print("Duplicate track IDs removed!")
print("Current Shape:", df.shape)
```

Create new columns

```python
# Convert milliseconds into minutes
df["duration_min"] = (
    df["duration_ms"] / 60000
)

# Create decade column
df["decade"] = (
    df["year"] // 10
) * 10

display(df.head())
```

Check cleaned data

```python
print("Missing values after cleaning:")
display(df.isnull().sum())
```

```python
print("Duplicate rows after cleaning:")
print(df.duplicated().sum())
```

```python
print("Final cleaned dataset shape:")
print(df.shape)
```

Task 5.1 — Build Predictive Models to Forecast Song Popularity
Answer


Machine-learning models are used to predict the popularity of songs based on their audio features. In this project, Linear Regression and Decision Tree Regression are used.

```python
# Import machine-learning libraries

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
# Select input features

features = [
    "danceability",
    "energy",
    "tempo",
    "valence",
    "acousticness",
    "instrumentalness",
    "liveness",
    "loudness",
    "speechiness",
    "duration_ms",
    "explicit"
]

X = df[features].copy()

# Convert Boolean values into integers

X["explicit"] = X["explicit"].astype(int)

# Target variable

y = df["popularity"]

# Split the data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# Train Linear Regression model

linear_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

linear_model.fit(
    X_train,
    y_train
)

linear_predictions = linear_model.predict(
    X_test
)

print("Linear Regression model trained successfully.")
```

```python
# Train Decision Tree model

tree_model = DecisionTreeRegressor(
    max_depth=10,
    random_state=42
)

tree_model.fit(
    X_train,
    y_train
)

tree_predictions = tree_model.predict(
    X_test
)

print("Decision Tree model trained successfully.")
```

Task 5.2: Evaluate the performance of different models (e.g., linear regression, decision trees).

```python
# Evaluate Linear Regression

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("LINEAR REGRESSION")
print("MAE:", linear_mae)
print("RMSE:", linear_rmse)
print("R2 Score:", linear_r2)
```

```python
# Evaluate Decision Tree

tree_mae = mean_absolute_error(
    y_test,
    tree_predictions
)

tree_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        tree_predictions
    )
)

tree_r2 = r2_score(
    y_test,
    tree_predictions
)

print("DECISION TREE")
print("MAE:", tree_mae)
print("RMSE:", tree_rmse)
print("R2 Score:", tree_r2)
```

```python
# Compare model performance

model_comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree"
    ],
    "MAE": [
        linear_mae,
        tree_mae
    ],
    "RMSE": [
        linear_rmse,
        tree_rmse
    ],
    "R2 Score": [
        linear_r2,
        tree_r2
    ]
})

display(model_comparison)
```

```python
# Plot model errors

model_comparison.set_index(
    "Model"
)[["MAE", "RMSE"]].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Comparison of Model Errors")
plt.xlabel("Model")
plt.ylabel("Error")
plt.xticks(rotation=0)
plt.show()
```

```python
# Actual vs predicted popularity

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    linear_predictions,
    alpha=0.4
)

plt.xlabel("Actual Popularity")
plt.ylabel("Predicted Popularity")
plt.title("Actual vs Predicted Popularity")

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)

plt.show()
```

Task 5.3: Fine-tune models to improve accuracy.

Answer

Fine-tuning is the process of testing different hyperparameter values to improve model performance. For the Decision Tree model, parameters such as max_depth, min_samples_split, and min_samples_leaf are tested.

GridSearchCV uses cross-validation to find the best combination of parameters. The tuned model is then evaluated on the test dataset.

```python
# Import GridSearchCV

from sklearn.model_selection import GridSearchCV
```

```python
# Define hyperparameter values

param_grid = {
    "max_depth": [5, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5]
}
```

```python
# Perform Grid Search

grid_search = GridSearchCV(
    estimator=DecisionTreeRegressor(
        random_state=42
    ),
    param_grid=param_grid,
    cv=3,
    scoring="neg_mean_squared_error",
    n_jobs=-1
)

grid_search.fit(
    X_train,
    y_train
)

print("Best Parameters:")
print(grid_search.best_params_)
```

```python
# Train and evaluate the tuned model

best_tree = grid_search.best_estimator_

best_predictions = best_tree.predict(
    X_test
)

best_mae = mean_absolute_error(
    y_test,
    best_predictions
)

best_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        best_predictions
    )
)

best_r2 = r2_score(
    y_test,
    best_predictions
)

print("TUNED DECISION TREE")
print("MAE:", best_mae)
print("RMSE:", best_rmse)
print("R2 Score:", best_r2)
```

```python
# Compare original and tuned Decision Tree

tuned_comparison = pd.DataFrame({
    "Model": [
        "Original Decision Tree",
        "Tuned Decision Tree"
    ],
    "MAE": [
        tree_mae,
        best_mae
    ],
    "RMSE": [
        tree_rmse,
        best_rmse
    ],
    "R2 Score": [
        tree_r2,
        best_r2
    ]
})

display(tuned_comparison)
```

```python
# Compare original and tuned Decision Tree

tuned_comparison = pd.DataFrame({
    "Model": [
        "Original Decision Tree",
        "Tuned Decision Tree"
    ],
    "MAE": [
        tree_mae,
        best_mae
    ],
    "RMSE": [
        tree_rmse,
        best_rmse
    ],
    "R2 Score": [
        tree_r2,
        best_r2
    ]
})

display(tuned_comparison)
```

```python
feature_importance = pd.DataFrame({
    "Feature": features,
    "Importance": best_tree.feature_importances_
}).sort_values("Importance", ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance in Tuned Decision Tree")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()
```
