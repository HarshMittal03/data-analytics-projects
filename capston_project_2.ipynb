FILL NULL VALUE

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
```

```python
data = df.copy()

num_cols = ["average_cost_for_two","city_id", "latitude","longitude","country_id","price_range","aggregate_rating","votes","photo_count","opentable_support","delivery","takeaway"]
cat_cols = ["locality_verbose","address", "city", "locality", "cuisines","currency","highlights","rating_text", "timings"]

for col in num_cols:
    data[col] = data[col].fillna(data[col].median())

for col in cat_cols:
    data[col] = data[col].fillna(data[col].mode()[0])

print("missing values:", data.isnull().sum().sum())
```

Rating vs Votes

```python
sample = df.sample(min(15000, len(df)), random_state=42)
plt.figure(figsize=(10,6))
plt.scatter(sample["votes"], sample["aggregate_rating"], s=6, alpha=.25)
plt.xlabel("Votes"); plt.ylabel("Aggregate Rating")
plt.title("Rating vs Number of Votes")
```

What this visualization shows: Tests whether restaurants receiving more customer votes tend to have different ratings.

Log Votes vs Rating

```python
sample = df[df["votes"] > 0].sample(min(15000, (df["votes"] > 0).sum()), random_state=42)
plt.figure(figsize=(10,6))
plt.scatter(np.log1p(sample["votes"]), sample["aggregate_rating"], s=6, alpha=.25)
plt.xlabel("log(1 + Votes)"); plt.ylabel("Aggregate Rating")
plt.title("Log Votes vs Rating")
plt.show()
```

What this visualization shows: Log transformation reduces the effect of extremely popular restaurants and makes the vote-rating pattern easier to inspect.

Photo Count Distribution

```python
limit = df["photo_count"].quantile(.99)
plt.figure(figsize=(10,6))
df["photo_count"].clip(upper=limit).plot(kind="hist", bins=30)
plt.xlabel("Photo Count"); plt.ylabel("Restaurants")
plt.title("Photo Count Distribution (99th Percentile Capped)")
plt.show()
```

What this visualization shows: Displays the distribution of restaurant photo activity while limiting extreme outliers for readability.

Rating by Cuisine Variety

```python
tmp = df.copy()
tmp["cuisine_count"] = tmp["cuisines"].fillna("").apply(     lambda x: len([z.strip() for z in str(x).split(",") if z.strip()]) )
plt.figure(figsize=(10,6))
tmp.boxplot(column="aggregate_rating", by="cuisine_count")
plt.suptitle("")
plt.title("Rating by Number of Cuisines Offered")
plt.xlabel("Cuisine Count");
plt.ylabel("Aggregate Rating")
plt.show()
```

What this visualization shows: Tests whether offering more cuisine types is associated with a different rating distribution.

Average Rating of Popular Cuisines

```python
tmp = df[["cuisines","aggregate_rating"]].dropna().copy()
tmp["cuisine"] = tmp["cuisines"].str.split(",")
tmp = tmp.explode("cuisine")
tmp["cuisine"] = tmp["cuisine"].str.strip()
popular = tmp["cuisine"].value_counts().head(15).index
avg = tmp[tmp["cuisine"].isin(popular)].groupby("cuisine")["aggregate_rating"].mean().sort_values()
plt.figure(figsize=(10,6))
avg.plot(kind="barh")
plt.xlabel("Average Rating"); plt.ylabel("Cuisine")
plt.title("Average Rating of Popular Individual Cuisines")
plt.show()
```

What this visualization shows: Compares the average ratings of the most frequently listed individual cuisines.

Delivery vs Median Votes

```python
med = df.groupby("delivery")["votes"].median()
plt.figure(figsize=(8,5))
med.plot(kind="bar")
plt.xlabel("Delivery"); plt.ylabel("Median Votes")
plt.title("Median Votes by Delivery Availability")
plt.xticks(rotation=0)
plt.show()
```

What this visualization shows: Compares typical customer voting activity between delivery categories.

Delivery + Takeaway Combination

```python
combo = df.groupby(["delivery","takeaway"]).size()
combo.index = [f"Delivery={a}, Takeaway={b}" for a,b in combo.index]
plt.figure(figsize=(10,5))
combo.plot(kind="bar")
plt.xlabel("Service Combination"); plt.ylabel("Restaurants")
plt.title("Restaurant Count by Delivery + Takeaway Combination")
plt.xticks(rotation=30, ha="right")
plt.show()
```

What this visualization shows: Shows the most common combinations of the two customer service options.

Top 5 Establishment Types Within Top 5 Cities

```python
top5c = df["city"].value_counts().head(5).index
tmp_establishment = df["establishment"].dropna().astype(str).str.strip('[]').str.split(', ').explode()
df['establishment_clean'] = tmp_establishment.str.strip("'").str.strip()
top5e = df["establishment_clean"].value_counts().head(5).index
mix = (df[df["city"].isin(top5c) & df["establishment_clean"].isin(top5e)]
       .groupby(["city","establishment_clean"]).size().reset_index(name="count"))

plt.figure(figsize=(11,6))
sns.barplot(data=mix, x="city", y="count", hue="establishment_clean", palette="Set2")
plt.title("Top 5 Establishment Types Within Top 5 Cities")
plt.xticks(rotation=15)
plt.legend(title="Establishment", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.show()
```
